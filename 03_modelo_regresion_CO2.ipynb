# Modelo de regresión lineal de las emisiones mundiales de CO₂

## 1. Introducción
La regresión lineal es una herramienta estadística que permite analizar la relación entre dos variables y representar su comportamiento mediante una línea de tendencia. Su aplicación en el estudio de las emisiones de dióxido de carbono (CO₂) permite examinar cómo han evolucionado las emisiones mundiales a través del tiempo.

En esta etapa del proyecto se desarrollará un modelo de regresión lineal utilizando los datos históricos de las emisiones mundiales de CO₂. A partir de este modelo se realizarán predicciones y se evaluará su desempeño mediante diferentes métricas estadísticas, con el propósito de comprender su capacidad para representar la tendencia de los datos.

## 2. Objetivos

### 2.1. Objetivo general
Construir y evaluar un modelo de regresión lineal que permita analizar la tendencia histórica de las emisiones mundiales de CO₂ a partir de los registros disponibles.

### 2.2. Objetivos específicos
* Preparar los datos para la construcción del modelo de regresión lineal.
* Identificar la relación entre el año y las emisiones mundiales de CO₂.
* Entrenar un modelo de regresión lineal utilizando los datos históricos.
* Evaluar el desempeño del modelo mediante métricas estadísticas.
* Comparar los valores reales con las predicciones obtenidas e interpretar los resultados.

## 3. Importación de librerías
Para desarrollar el modelo se utilizaron diferentes librerías de Python que permiten manipular datos, realizar cálculos matemáticos, construir modelos estadísticos y representar gráficamente los resultados.

* **Pandas:** permite organizar y manipular la información.
* **NumPy:** facilita las operaciones matemáticas.
* **Matplotlib:** permite elaborar gráficos.
* **Scikit-learn:** proporciona herramientas para construir y evaluar modelos de aprendizaje automático, como la regresión lineal.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

plt.style.use('default')
print("Librerías importadas correctamente")
```

## 4. Carga de la base de datos
Para desarrollar el modelo se utilizó la base de datos limpia generada durante la primera etapa del proyecto. Esta contiene 276 registros de las emisiones mundiales de CO₂, correspondientes al periodo comprendido entre 1750 y 2025.

La información se cargó en Python para preparar las variables necesarias y realizar el proceso de entrenamiento y evaluación del modelo.

```python
from google.colab import files
archivo = files.upload()
nombre_archivo = next(iter(archivo))
datos = pd.read_csv(nombre_archivo)
display(datos.head())
```

```python
print("Cantidad de registros:", len(datos))
print("Columnas:", list(datos.columns))
print("Periodo:", datos['anio'].min(), "-", datos['anio'].max())
```

## 5. Preparación de los datos
Para construir el modelo de regresión lineal se definieron dos variables principales. El año se estableció como variable independiente (X), mientras que las emisiones mundiales de CO₂ se utilizaron como variable dependiente (Y).

Posteriormente, los datos se dividieron en dos conjuntos: entrenamiento y prueba. Se utilizó el 80 % de los registros para entrenar el modelo y el 20 % restante para evaluar su desempeño.

La división se realizó conservando el orden cronológico de los datos, de manera que el modelo fuera evaluado con años posteriores a los utilizados durante su entrenamiento.

```python
X = datos[['anio']]
y = datos['emisiones_co2_toneladas']
punto_corte = int(len(datos) * 0.8)
X_train = X.iloc[:punto_corte]
X_test = X.iloc[punto_corte:]
y_train = y.iloc[:punto_corte]
y_test = y.iloc[punto_corte:]

print("Registros de entrenamiento:", len(X_train))
print("Registros de prueba:", len(X_test))
```

## 6. Construcción y entrenamiento del modelo
Se utilizó el algoritmo de regresión lineal de la librería Scikit-learn para establecer una relación matemática entre el año y las emisiones mundiales de CO₂.

El modelo se entrenó utilizando el conjunto de datos destinado para este propósito. Durante este proceso se identificaron los parámetros de la ecuación lineal, correspondientes a la pendiente y el intercepto, que permiten representar la tendencia de las emisiones a través del tiempo.

```python
modelo = LinearRegression()
modelo.fit(X_train, y_train)
print("Modelo entrenado correctamente")
print("Pendiente:", modelo.coef_[0])
print("Intercepto:", modelo.intercept_)
```

## 7. Predicciones del modelo
Una vez entrenado el modelo de regresión lineal, se realizaron predicciones utilizando los años correspondientes al conjunto de prueba.

Estas predicciones permiten estimar las emisiones mundiales de CO₂ y compararlas con los valores reales registrados en la base de datos, con el propósito de identificar las diferencias entre ambos resultados.

```python
predicciones = modelo.predict(X_test)
resultados = pd.DataFrame({
    'Anio': X_test['anio'].values,
    'Emisiones_reales': y_test.values,
    'Emisiones_predichas': predicciones
})
display(resultados.head())
```

## 8. Evaluación del modelo
Para evaluar el desempeño del modelo se utilizaron tres métricas estadísticas que permiten conocer qué tan cercanas son las predicciones a los valores reales.

* **R² (coeficiente de determinación):** permite conocer qué proporción de la variabilidad de los datos es explicada por el modelo.
* **MAE (error absoluto medio):** representa el promedio de las diferencias absolutas entre los valores reales y los predichos.
* **RMSE (raíz del error cuadrático medio):** mide el error de las predicciones y penaliza en mayor medida los errores grandes.

Estas métricas permiten analizar la capacidad del modelo para representar el comportamiento de las emisiones mundiales de CO₂.

```python
r2 = r2_score(y_test, predicciones)
mae = mean_absolute_error(y_test, predicciones)
rmse = np.sqrt(mean_squared_error(y_test, predicciones))

print("Coeficiente de determinación (R²):", r2)
print("Error absoluto medio (MAE):", mae)
print("Raíz del error cuadrático medio (RMSE):", rmse)
```

## 9. Comparación entre valores reales y predichos
Se elaboró una representación gráfica para comparar las emisiones reales con las estimaciones realizadas por el modelo de regresión lineal.

Esta visualización permite observar las diferencias entre ambos valores y analizar qué tan bien representa el modelo la tendencia de las emisiones durante el periodo de prueba.

```python
plt.figure(figsize=(12, 6))
plt.plot(
    X_test['anio'],
    y_test,
    label='Emisiones reales'
)
plt.plot(
    X_test['anio'],
    predicciones,
    label='Emisiones predichas',
    linestyle='--'
)
plt.title('Comparación entre emisiones reales y predichas')
plt.xlabel('Año')
plt.ylabel('Emisiones de CO₂ (toneladas)')
plt.legend()
plt.grid(True)
plt.show()
```

## 10. Interpretación de resultados
En esta sección se analizarán los resultados obtenidos durante la construcción y evaluación del modelo de regresión lineal.

Se tendrán en cuenta los valores de las métricas estadísticas, las diferencias entre las emisiones reales y las predichas, y el comportamiento de la línea de tendencia.

También se considerarán las limitaciones del modelo, teniendo en cuenta que la regresión lineal representa una tendencia general y no necesariamente refleja todos los cambios que pueden presentarse en las emisiones a lo largo del tiempo.

El modelo permite identificar la tendencia histórica de las emisiones mundiales de CO₂ y establecer su relación con el paso de los años. Las métricas R², MAE y RMSE permiten evaluar qué tan precisas son las predicciones y cuánto se alejan de los valores reales.

Por su parte, la gráfica permite comparar las emisiones registradas con las estimadas por el modelo, identificando las diferencias entre ambas. Estas variaciones muestran que la regresión lineal puede representar una tendencia general, pero no siempre logra ajustarse a los cambios que presentan las emisiones.

Finalmente, es importante tener en cuenta que las emisiones mundiales dependen de factores como la industrialización, el crecimiento económico y el consumo de combustibles fósiles. Por esta razón, utilizar únicamente el año como variable independiente limita la precisión de las predicciones.

```python
print("Pendiente:", modelo.coef_[0])
print("Intercepto:", modelo.intercept_)
print("R²:", r2)
print("MAE:", mae)
print("RMSE:", rmse)
```
