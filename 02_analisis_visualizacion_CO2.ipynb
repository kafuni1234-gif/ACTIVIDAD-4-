# **Análisis y visualización de las emisiones mundiales de CO₂**

## **1. Introducción**

El análisis de datos permite identificar comportamientos y tendencias que ayudan a comprender diferentes problemáticas ambientales. En este caso, se estudiará la evolución histórica de las emisiones mundiales de dióxido de carbono (CO₂), utilizando la información organizada durante la primera etapa del proyecto.

Mediante herramientas estadísticas y representaciones gráficas en Python, se busca comprender cómo han cambiado las emisiones a lo largo del tiempo y reconocer los principales comportamientos presentes en los datos.

## **2. Objetivos**

 2.1. Objetivo general

Analizar el comportamiento histórico de las emisiones mundiales de CO₂ mediante herramientas estadísticas y gráficas, con el propósito de identificar tendencias y cambios a lo largo del periodo estudiado.

2.2. Objetivos específicos

- Calcular medidas estadísticas descriptivas de las emisiones mundiales de CO₂.
- Representar gráficamente la evolución de las emisiones a través del tiempo.
- Identificar tendencias y variaciones en el comportamiento de los datos.
- Interpretar los resultados obtenidos mediante el análisis estadístico y gráfico.
