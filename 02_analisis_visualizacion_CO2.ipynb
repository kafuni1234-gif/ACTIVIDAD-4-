## **6. Evolución histórica de las emisiones mundiales de CO₂**

Se elaboró un gráfico de líneas para representar la evolución de las emisiones mundiales de CO₂ entre 1750 y 2025.

Esta representación permite observar los cambios que han presentado las emisiones a lo largo del tiempo, identificar periodos de crecimiento y reconocer posibles disminuciones.

```python
plt.figure(figsize=(12,6))

plt.plot(
    datos['anio'],
    datos['emisiones_co2_toneladas'],
    color='blue'
)

plt.title('Evolución de las emisiones mundiales de CO₂')
plt.xlabel('Año')
plt.ylabel('Emisiones de CO₂ (toneladas)')
plt.grid(True)

plt.show()
```

La gráfica muestra la evolución de las emisiones mundiales de dióxido de carbono (CO₂) entre los años 1750 y 2025. Se observa que durante los primeros años del periodo analizado las emisiones permanecieron en niveles relativamente bajos y con incrementos graduales. Sin embargo, a partir del siglo XX, especialmente después de 1950, las emisiones experimentaron un crecimiento acelerado.

La tendencia ascendente observada está asociada al aumento de las actividades industriales, el crecimiento económico y el uso intensivo de combustibles fósiles como fuente principal de energía. Aunque se presentan ligeras disminuciones en algunos años, la tendencia general evidencia un incremento sostenido de las emisiones mundiales de CO₂ a lo largo del tiempo.

Los mayores valores de emisiones se registran en los años más recientes del conjunto de datos, lo que refleja la creciente demanda energética y el impacto de las actividades humanas sobre el medio ambiente.

## **7. Relación entre el año y las emisiones de CO₂**
Con el fin de analizar la relación existente entre el paso del tiempo y las emisiones mundiales de CO₂, se construyó un gráfico de dispersión. Este tipo de representación permite observar la distribución de los datos e identificar posibles tendencias o patrones a lo largo del periodo estudiado.

Cada punto representa las emisiones registradas en un año específico, facilitando la identificación visual de cambios significativos en el comportamiento de las emisiones.

```python
plt.figure(figsize=(12,6))

plt.scatter(
    datos['anio'],
    datos['emisiones_co2_toneladas'],
    color='darkred',
    alpha=0.7
)

plt.title('Relación entre el año y las emisiones mundiales de CO₂')
plt.xlabel('Año')
plt.ylabel('Emisiones de CO₂ (toneladas)')
plt.grid(True)

plt.show()
```

El gráfico de dispersión permite observar la relación existente entre el año y las emisiones mundiales de CO₂. Se identifica una tendencia positiva, ya que las emisiones aumentan conforme avanza el tiempo.

Durante el periodo comprendido entre 1750 y finales del siglo XIX, las emisiones permanecieron relativamente bajas y estables. Sin embargo, a partir del siglo XX se evidencia un crecimiento progresivo que se intensifica notablemente después de 1950.

La concentración de puntos en la parte superior derecha del gráfico muestra que los años más recientes registran los niveles más altos de emisiones. En conjunto, los resultados indican una tendencia creciente de las emisiones mundiales de CO₂ a lo largo del tiempo.

7.1. Tendencia de las emisiones mundiales de CO₂

Con el fin de visualizar con mayor claridad la tendencia general de las emisiones mundiales de CO₂, se incorpora una línea de tendencia mediante una regresión lineal simple. Esta representación permite identificar el comportamiento global de los datos y verificar si existe una tendencia creciente, decreciente o estable a lo largo del tiempo.

```python
plt.figure(figsize=(12,6))

sns.regplot(
    x='anio',
    y='emisiones_co2_toneladas',
    data=datos,
    scatter_kws={'alpha':0.6, 'color':'darkred'},
    line_kws={'color':'blue', 'linewidth':2}
)

plt.title('Tendencia de las emisiones mundiales de CO₂')
plt.xlabel('Año')
plt.ylabel('Emisiones de CO₂ (toneladas)')
plt.grid(True)

plt.show()
```

La gráfica de tendencia permite observar la relación general entre el año y las emisiones mundiales de CO₂ mediante una línea de regresión lineal. La pendiente positiva de la línea indica que existe una relación directa entre el transcurso del tiempo y el aumento de las emisiones.

Aunque la línea de tendencia resume el comportamiento general de los datos, se observa que el crecimiento real de las emisiones no es completamente lineal. Durante los primeros años del periodo analizado las emisiones fueron relativamente bajas, mientras que a partir de mediados del siglo XX se produjo un incremento mucho más acelerado.

La separación entre algunos puntos y la línea de tendencia evidencia que las emisiones han crecido a un ritmo cada vez mayor, especialmente en las últimas décadas. En consecuencia, puede concluirse que las emisiones mundiales de CO₂ presentan una tendencia claramente creciente a lo largo del tiempo.
