## 5. Descarga de la base de datos

Para obtener la información necesaria para el proyecto, se descargó la base de datos de emisiones anuales de CO₂ desde la plataforma **Our World in Data**.

El archivo contiene información histórica de diferentes países, regiones y agrupaciones mundiales. Los datos se cargaron directamente en Python mediante la función `read_csv()` de Pandas, lo que permitió organizarlos en una estructura tabular para realizar su exploración y posterior procesamiento.

```python
# Descargar la base de datos de emisiones de CO2

url = "https://ourworldindata.org/grapher/annual-co2-emissions-per-country.csv?v=1&csvType=full&useColumnShortNames=true"

datos = pd.read_csv(
    url,
    storage_options={
        'User-Agent': 'Our World In Data data fetch/1.0'
    }
)

print("Datos descargados correctamente")
```

## 7. Identificación de valores faltantes

Se revisó la presencia de valores faltantes en las diferentes columnas de la base de datos, con el propósito de identificar posibles vacíos de información que pudieran afectar el análisis.

Durante esta revisión se encontraron **3.330 valores faltantes** en la columna correspondiente al código de identificación. Sin embargo, no se eliminaron estos registros, ya que el estudio se enfocará posteriormente en seleccionar únicamente la información mundial.

## 8. Selección de los datos mundiales

La base de datos original contiene información de **252 entidades diferentes**, entre las que se encuentran países, continentes, regiones y agrupaciones económicas.

Para mantener un enfoque uniforme en el proyecto, se seleccionaron únicamente los registros identificados como **World**, que representan las emisiones mundiales de CO₂.

Esta selección permite estudiar el comportamiento global de las emisiones a lo largo del tiempo, sin mezclar los registros de países individuales con los datos de regiones o agrupaciones.

```python
# Filtrar únicamente los registros correspondientes al mundo

datos_mundo = datos[datos['entity'] == 'World'].copy()

# Mostrar los primeros registros

datos_mundo.head()
```

```python
# Revisar la cantidad de registros mundiales

datos_mundo.shape
```

```python
# Revisar el periodo de tiempo disponible

print("Año inicial:", datos_mundo['year'].min())
print("Año final:", datos_mundo['year'].max())
```

## 9. Revisión y limpieza de los datos seleccionados

Después de seleccionar los registros correspondientes a la entidad mundial (**World**), se verificó el periodo histórico disponible, la existencia de valores faltantes y la presencia de años duplicados.

Esta revisión permitió comprobar que la información seleccionada no presentaba valores faltantes ni registros repetidos por año. Posteriormente, los datos se organizaron cronológicamente para facilitar su análisis.

```python
# Revisar valores faltantes en los datos mundiales

datos_mundo.isnull().sum()
```

```python
# Revisar si existen años repetidos

datos_mundo['year'].duplicated().sum()
```

```python
# Ordenar los registros por año

datos_mundo = datos_mundo.sort_values(by='year').reset_index(drop=True)

# Mostrar los primeros registros

datos_mundo.head()
```

## 10. Preparación de la base de datos definitiva

Para facilitar el desarrollo de las siguientes etapas del proyecto, se conservaron únicamente las variables necesarias para estudiar la evolución de las emisiones mundiales: **el año y la cantidad de emisiones de CO₂**.

Las columnas se renombraron para facilitar su interpretación y la información se guardó en un archivo CSV que podrá ser utilizado por los otros integrantes del grupo.

```python
# Crear la base de datos definitiva

base_limpia = datos_mundo[
    ['year', 'emissions_total']
].copy()

# Cambiar el nombre de las columnas para facilitar su interpretación

base_limpia = base_limpia.rename(columns={
    'year': 'anio',
    'emissions_total': 'emisiones_co2_toneladas'
})

# Mostrar los primeros registros

base_limpia.head()
```

```python
# Exportar la base de datos limpia

base_limpia.to_csv(
    'emisiones_co2_mundiales_limpias.csv',
    index=False,
    encoding='utf-8'
)

print("Archivo CSV creado correctamente")
```

```python
# Verificar las dimensiones de la base final

print("Cantidad de registros:", len(base_limpia))
print("Columnas:", list(base_limpia.columns))
```
